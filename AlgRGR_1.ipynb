# Часть 1 - Теория и Модель $\mathrm{SO_2}$

 ## Основные понятия

 - ***Операция симметрии*** — преобразование, после которого молекула совпадает с собой. При этом одинаковые атомы могут поменяться местами.

 - ***Элемент симметрии*** — геометрический объект, относительно которого выполняется операция. Например, у $\mathrm{SO_2}$ прямая, делящая угол O–S–O пополам, служит осью симметрии. Если повернуть молекулу вокруг этой прямой на $180^\circ$, атом серы останется на месте, а два атома кислорода поменяются местами. Прямая — это элемент симметрии, а поворот — операция.

 - ***Точечная группа молекулы*** объединяет все её операции симметрии. Эти операции сохраняют хотя бы одну общую неподвижную точку. В модели $\mathrm{SO_2}$ такой точкой служит положение атома серы.



## Операции точечной симметрии

При определении точечной группы рассматривают преобразования, после которых каждый атом занимает исходное положение атома того же химического элемента.

| Обозначение | Операция | Как она действует |
| :---: | :---: |:---: |
| $E$ | Тождественная операция | Каждый атом остаётся в исходном положении. Эта операция входит в любую точечную группу. |
| $C_n$ | Поворот вокруг оси | Молекулу поворачивают на угол $360^\circ/n$. Например, $C_2$ означает поворот на $180^\circ$, а $C_3$ — на $120^\circ$. |
| $\sigma$ | Отражение в плоскости | Атом переходит в зеркальную точку по другую сторону плоскости на том же расстоянии от неё. Атомы в самой плоскости остаются на месте. |
| $i$ | Инверсия относительно центра | Если центр инверсии выбран за начало координат, то $(x,y,z)\mapsto(-x,-y,-z)$. |
| $S_n$ | Поворотно-зеркальная операция | Сначала выполняют поворот на $360^\circ/n$, затем отражение в плоскости, перпендикулярной оси поворота. Совпадение молекулы с исходной проверяют после обоих шагов. |  

**Повторные вращения:** Запись $C_n^k$ означает, что поворот $C_n$ выполнили $k$ раз вокруг одной оси. Например, в PCl₃ операции $C_3$ и $C_3^2$ поворачивают молекулу на $120^\circ$ и $240^\circ$ соответственно. После третьего поворота каждый атом возвращается в исходное положение, поэтому $C_3^3=E$. Так одна геометрическая ось может давать несколько различных операций.

### Как различать зеркальные плоскости

Главной называют ось вращения наибольшего порядка. Положение зеркальной плоскости описывают относительно этой оси.

- $\sigma_v$ — вертикальная плоскость. Она содержит главную ось. Например, у $\mathrm{SO_2}$ главная ось направлена вдоль $z$, поэтому обе плоскости $xz$ и $yz$ вертикальные. Первая совпадает с плоскостью молекулы, вторая меняет атомы кислорода местами.

- $\sigma_h$ — горизонтальная плоскость. Она перпендикулярна главной оси. У $\mathrm{PF_5}$ такая плоскость проходит через $P$ и три экваториальных атома $F$. При отражении два осевых атома $F$ меняются местами.

- $\sigma_d$ — диагональная плоскость. Она содержит главную ось и делит пополам угол между двумя соседними осями $C_2$, перпендикулярными главной.

### Как определить точечную группу

1. Зафиксировать пространственную форму молекулы. Найти главную ось $C_n$ и указать, какие атомы переставляются при повороте.

2. Проверить, есть ли $n$ осей $C_2$, перпендикулярных главной оси. Для рассматриваемых молекул это позволяет различить семейства $C_n$ и $D_n$.

3. Найти зеркальные плоскости, проверить наличие центра инверсии и поворотно-зеркальных операций. Для каждого преобразования убедиться, что атомы переходят в положения атомов того же элемента.

4. Выписать все различные операции, включая $E$. Их количество даёт порядок группы.

Эту схему применим к молекуле $\mathrm{SO_2}$. Сначала зададим координаты атомов, затем для каждой операции предскажем перестановку $O_1$ и $O_2$ и проверим её программно.

## Геометрия молекулы $\mathrm{SO_2}$

Молекула $\mathrm{SO_2}$ имеет угловую форму: в вершине угла O–S–O находится атом серы, а две связи S–O имеют одинаковую длину.

Поместим атом S в начало координат, а молекулу расположим в плоскости $xz$. Ось $z$ направим по биссектрисе угла O–S–O. Тогда координаты атомов можно записать как $S=(0,0,0)$, $O_1=(a,0,b)$ и $O_2=(-a,0,b)$, где $a,b>0$.

В схематической модели примем длину связи S–O за единицу, а угол O–S–O — равным $120^\circ$. Метки $O_1$ и $O_2$ помогут проследить, как при поворотах и отражениях меняются местами два одинаковых атома кислорода.

```python
import numpy as np
import plotly.graph_objects as go

half_angle = np.deg2rad(60)
a, b = np.sin(half_angle), np.cos(half_angle)

atoms = {
    "S":  np.array([0.0, 0.0, 0.0]),
    "O₁": np.array([a,   0.0, b]),
    "O₂": np.array([-a,  0.0, b]),
}

fig = go.Figure()

bonds = np.array([atoms["O₁"], atoms["S"], atoms["O₂"]])
fig.add_trace(go.Scatter3d(
    x=bonds[:, 0], y=bonds[:, 1], z=bonds[:, 2],
    mode="lines", line=dict(color="gray", width=6),
    showlegend=False,
))

for label, (x, y, z) in atoms.items():
    fig.add_trace(go.Scatter3d(
        x=[x], y=[y], z=[z], mode="markers+text",
        text=[label], textposition="top center",
        marker=dict(size=10, color="goldenrod" if label == "S" else "crimson"),
        showlegend=False,
    ))

fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, 0], z=[-0.45, 1.15],
    mode="lines", line=dict(color="black", width=5, dash="dash"),
    name="Ось C₂",
))

u, v = np.meshgrid([-1.1, 1.1], [-0.4, 1.05])
zero = np.zeros_like(u)
for name, x, y, color in [
    ("Плоскость xz", u, zero, "royalblue"),
    ("Плоскость yz", zero, u, "seagreen"),
]:
    fig.add_trace(go.Surface(
        x=x, y=y, z=v, surfacecolor=zero,
        colorscale=[[0, color], [1, color]],
        opacity=0.18, showscale=False, showlegend=True, name=name,
        hoverinfo="skip",
    ))

fig.update_layout(
    title="SO₂: ось C₂ и две плоскости симметрии",
    scene=dict(
        xaxis_title="x", yaxis_title="y", zaxis_title="z",
        aspectmode="data",
        camera=dict(eye=dict(x=1.5, y=2.0, z=1.1)),
    ),
    height=550,
)
fig.show()
```

## Операции симметрии $\mathrm{SO_2}$

| Операция | Как меняются координаты | Что происходит с атомами |
| :---: | :---: | :---: |
| $E$ - тождественная операция | $(x,y,z)\to(x,y,z)$ | Все атомы остаются на своих местах |
| $C_2(z)$ - поворот на $180^\circ$ вокруг оси $z$  | $(x,y,z)\to(-x,-y,z)$ | $O_1$ и $O_2$ меняются местами, $S$ остаётся на месте |
| $\sigma_{xz}$ - отражение в плоскости молекулы | $(x,y,z)\to(x,-y,z)$ | Каждый атом остаётся на месте, поскольку для всех трёх $y=0$ |
| $\sigma_{yz}$ - отражение в плоскости, делящей угол O-S-O пополам | $(x,y,z)\to(-x,y,z)$ | $O_1$ и $O_2$ меняются местами, $S$ остаётся на месте |  

***Вывод:***  У молекулы одна ось вращения второго порядка и две зеркальные плоскости, содержащие эту ось. Такие плоскости называют вертикальными и обозначают $\sigma_v$. Точечная группа $\mathrm{SO_2}$ — $C_{2v}$, её порядок равен 4.

 ### Программная проверка поворота $C_2(z)$ - на $180^\circ$ вокруг оси $z$

```python
import pandas as pd

# Поворот вокруг z на 180° меняет знаки x и y и сохраняет z.
C2_z = np.array([
    [-1,  0,  0],
    [ 0, -1,  0],
    [ 0,  0,  1],
])

rows = []
for label, before in atoms.items():
    after = C2_z @ before

    # Находим, на чьё исходное место попал атом.
    target = next(
        (name for name, position in atoms.items()
         if np.allclose(after, position, atol=1e-8, rtol=0)),
        "нет совпадения",
    )

    rows.append({
        "Атом": label,
        "До поворота (x, y, z)": tuple(before.round(3).tolist()),
        "После поворота (x, y, z)": tuple(after.round(3).tolist()),
        "На чьём месте оказался": target,
    })

pd.DataFrame(rows)
```

### Действие операций симметрии на 3D-модель $\mathrm{SO_2}$

Можно выбрать операцию в выпадающем списке. Серым будет показано исходное положение атомов, красным - положение после преобразования.

```python
import numpy as np
import plotly.graph_objects as go

labels = ["O₁", "S", "O₂"]
before = np.array([atoms[label] for label in labels])

operations = {
    "E — без изменений": np.eye(3),
    "C₂(z) — поворот на 180°": np.diag([-1, -1, 1]),
    "σ(xz) — отражение в плоскости молекулы": np.diag([1, -1, 1]),
    "σ(yz) — отражение в плоскости yz": np.diag([-1, 1, 1]),
}

fig = go.Figure()

def add_model(points, name, color, size, text_position, visible=True):
    fig.add_scatter3d(
        x=points[:, 0], y=points[:, 1], z=points[:, 2],
        mode="lines+markers+text",
        text=[f"{name}: {label}" for label in labels],
        textposition=text_position,
        textfont=dict(color=color),
        marker=dict(size=size, color=color),
        line=dict(color=color, width=4),
        name=name,
        visible=visible,
    )

# Исходная модель видна при любом выборе.
add_model(before, "До", "dimgray", 12, "bottom center")

buttons = []
for index, (title, matrix) in enumerate(operations.items()):
    after = (matrix @ before.T).T
    add_model(after, "После", "crimson", 6, "top center", visible=(index == 0))

    # Показываем исходную модель и один выбранный результат.
    visible = [True] + [i == index for i in range(len(operations))]
    buttons.append(dict(label=title, method="update", args=[{"visible": visible}]))

# Добавляем на график ось C₂ и две зеркальные плоскости.
fig.add_scatter3d(
    x=[0, 0], y=[0, 0], z=[-0.4, 1.0],
    mode="lines",
    line=dict(color="black", width=5, dash="dash"),
    name="Ось C₂", visible=False,
)

u, v = np.meshgrid([-1.1, 1.1], [-0.4, 1.0])
zero = np.zeros_like(u)

for name, x, y, color in [
    ("Плоскость xz", u, zero, "royalblue"),
    ("Плоскость yz", zero, u, "seagreen"),
]:
    fig.add_surface(
        x=x, y=y, z=v, surfacecolor=zero,
        colorscale=[[0, color], [1, color]],
        opacity=0.2, showscale=False, showlegend=True,
        name=name, visible=False, hoverinfo="skip",
    )

# Порядок пунктов меню: E, C₂, σ(xz), σ(yz).
for index, button in enumerate(buttons):
    button["args"][0]["visible"].extend([
        index == 1,  # ось C₂
        index == 2,  # плоскость xz
        index == 3,  # плоскость yz
    ])

fig.update_layout(
    template="plotly_white",
    updatemenus=[dict(buttons=buttons, x=0, y=1.15, xanchor="left")],
    margin=dict(t=110, b=20),
    height=530,
    scene=dict(
        xaxis_title="x", yaxis_title="y", zaxis_title="z",
        xaxis=dict(range=[-1.3, 1.3]),
        yaxis=dict(range=[-0.8, 0.8]),
        zaxis=dict(range=[-0.4, 1.0]),
        aspectmode="data",
    ),
)

fig.show()
```

### Проверка совпадения модели $\mathrm{SO_2}$ для всех операций симметрии

```python
import pandas as pd

elements = {"S": "S", "O₁": "O", "O₂": "O"}
rows = []

for operation, matrix in operations.items():
    targets = {}

    for label, position in atoms.items():
        after = matrix @ position

        # Ищем исходное место атома того же химического элемента.
        matches = [
            name for name, original in atoms.items()
            if elements[name] == elements[label]
            and np.linalg.norm(after - original) < 1e-8
        ]
        targets[label] = matches[0] if len(matches) == 1 else "нет совпадения"

    # Каждое исходное место должно быть занято ровно один раз.
    passed = set(targets.values()) == set(atoms)

    rows.append({
        "Операция": operation,
        "Куда перешёл S": targets["S"],
        "Куда перешёл O₁": targets["O₁"],
        "Куда перешёл O₂": targets["O₂"],
        "Модель совпала": "Да" if passed else "Нет",
    })

pd.DataFrame(rows)
```

## PCl₃ и полярность

```python
import numpy as np
import plotly.graph_objects as go
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch
```

### Интерактивная 3D-модель $\mathrm{PCl_3}$ и выбор всех 6 операций

```python
R = 1.0
h = 0.65
a120 = np.deg2rad(120)
a240 = np.deg2rad(240)

atoms = {
    "P":   np.array([0.0, 0.0, h]),
    "Cl₁": np.array([R, 0.0, 0.0]),
    "Cl₂": np.array([R * np.cos(a120), R * np.sin(a120), 0.0]),
    "Cl₃": np.array([R * np.cos(a240), R * np.sin(a240), 0.0]),
}

def rot_z(deg):
    r = np.deg2rad(deg)
    return np.array([[np.cos(r), -np.sin(r), 0], [np.sin(r), np.cos(r), 0], [0, 0, 1]])

def refl_v(deg):
    r = np.deg2rad(2 * deg)
    return np.array([[np.cos(r), np.sin(r), 0], [np.sin(r), -np.cos(r), 0], [0, 0, 1]])

operations = {
    "E — без изменений": np.eye(3),
    "C₃¹(z) — поворот на 120°": rot_z(120),
    "C₃²(z) — поворот на 240°": rot_z(240),
    "σᵥ₁ — плоскость через P и Cl₁": refl_v(0),
    "σᵥ₂ — плоскость через P и Cl₂": refl_v(120),
    "σᵥ₃ — плоскость через P и Cl₃": refl_v(240),
}

labels = ["P", "Cl₁", "Cl₂", "Cl₃"]
bonds = [("P", "Cl₁"), ("P", "Cl₂"), ("P", "Cl₃")]

fig = go.Figure()

def add_model(pts_dict, name, color, size, text_position, visible=True):
    bx, by, bz = [], [], []
    for a1, a2 in bonds:
        bx.extend([pts_dict[a1][0], pts_dict[a2][0], None])
        by.extend([pts_dict[a1][1], pts_dict[a2][1], None])
        bz.extend([pts_dict[a1][2], pts_dict[a2][2], None])
    fig.add_scatter3d(
        x=bx, y=by, z=bz, mode="lines",
        line=dict(color=color, width=4), showlegend=False, visible=visible
    )
    pts = np.array([pts_dict[l] for l in labels])
    fig.add_scatter3d(
        x=pts[:, 0], y=pts[:, 1], z=pts[:, 2],
        mode="markers+text",
        text=[f"{name}: {label}" for label in labels],
        textposition=text_position,
        textfont=dict(color=color),
        marker=dict(size=size, color=color),
        name=name, visible=visible,
    )

# Исходная модель
add_model(atoms, "До", "dimgray", 12, "bottom center")

buttons = []
for index, (title, matrix) in enumerate(operations.items()):
    after_dict = {k: matrix @ v for k, v in atoms.items()}
    add_model(after_dict, "После", "crimson", 6, "top center", visible=(index == 0))

    visible = [True, True]
    for i in range(len(operations)):
        visible.extend([i == index, i == index])
    visible.extend([
        index in [1, 2],  # ось C3
        index == 3,       # sigma_v1
        index == 4,       # sigma_v2
        index == 5,       # sigma_v3
    ])
    buttons.append(dict(label=title, method="update", args=[{"visible": visible}]))

# Ось C3 и три плоскости симметрии
fig.add_scatter3d(
    x=[0, 0], y=[0, 0], z=[-0.3, 1.0],
    mode="lines", line=dict(color="black", width=5, dash="dash"),
    name="Ось C₃", visible=False,
)

r_g, z_g = np.meshgrid([-1.15, 1.15], [-0.25, 0.95])
zero = np.zeros_like(r_g)
for name, deg, color in [
    ("Плоскость σᵥ₁", 0, "royalblue"),
    ("Плоскость σᵥ₂", 120, "seagreen"),
    ("Плоскость σᵥ₃", 240, "darkorange"),
]:
    rad = np.deg2rad(deg)
    fig.add_surface(
        x=r_g * np.cos(rad), y=r_g * np.sin(rad), z=z_g,
        surfacecolor=zero, colorscale=[[0, color], [1, color]],
        opacity=0.2, showscale=False, showlegend=True,
        name=name, visible=False, hoverinfo="skip",
    )

fig.update_layout(
    template="plotly_white",
    updatemenus=[dict(buttons=buttons, x=0, y=1.15, xanchor="left")],
    margin=dict(t=110, b=20), height=530,
    scene=dict(
        xaxis_title="x", yaxis_title="y", zaxis_title="z",
        xaxis=dict(range=[-1.3, 1.3]),
        yaxis=dict(range=[-1.3, 1.3]),
        zaxis=dict(range=[-0.3, 1.0]),
        aspectmode="data",
    ),
)

fig.show()
```

### Таблица программной проверки всех 6 операций $\mathrm{PCl_3}$

```python
elements = {"P": "P", "Cl₁": "Cl", "Cl₂": "Cl", "Cl₃": "Cl"}
rows = []

for operation, matrix in operations.items():
    targets = {}
    for label, position in atoms.items():
        after = matrix @ position
        matches = [
            name for name, original in atoms.items()
            if elements[name] == elements[label]
            and np.linalg.norm(after - original) < 1e-8
        ]
        targets[label] = matches[0] if len(matches) == 1 else "нет совпадения"

    passed = set(targets.values()) == set(atoms)
    rows.append({
        "Операция": operation,
        "Куда перешёл P": targets["P"],
        "Куда перешёл Cl₁": targets["Cl₁"],
        "Куда перешёл Cl₂": targets["Cl₂"],
        "Куда перешёл Cl₃": targets["Cl₃"],
        "Модель совпала": "Да" if passed else "Нет",
    })

pd.DataFrame(rows)
```

## Вывод по части 1

Все четыре операции сохраняют модель $\mathrm{SO_2}$. Поворот на $180^\circ$ вокруг оси $z$ и отражение в плоскости $yz$ меняют местами атомы кислорода. Отражение в плоскости молекулы $xz$ оставляет каждый атом на месте. У $\mathrm{SO_2}$ одна ось второго порядка и две зеркальные плоскости, проходящие через неё. Точечная группа молекулы - $C_{2v}$, порядок группы - $4$.
